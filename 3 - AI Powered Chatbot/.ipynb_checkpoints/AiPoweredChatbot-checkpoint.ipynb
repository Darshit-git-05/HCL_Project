```python
import warnings
warnings.filterwarnings("ignore")

import os
import re
import string
import pickle
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix
)

print("Libraries imported successfully.")
```

```python
text_candidates = []

for col in df.columns:
    if df[col].dtype == "object":
        avg_length = df[col].astype(str).str.len().mean()
        text_candidates.append((col, avg_length))

text_candidates = sorted(
    text_candidates,
    key=lambda x: x[1],
    reverse=True
)

print("Possible text columns:")
for col, avg_length in text_candidates:
    print(col, "Average length:", round(avg_length, 2))
```

```python
# Detect likely label column
label_candidates = []

for col in df.columns:
    if df[col].dtype == "object":
        unique_count = df[col].nunique()

        if unique_count < len(df) * 0.5:
            label_candidates.append((col, unique_count))

print("Possible label columns:")

for col, count in label_candidates:
    print(col, "Unique values:", count)
```

```python
TEXT_COLUMN = "Question"
LABEL_COLUMN = "Intent"
```

```python
TEXT_COLUMN = "text"
LABEL_COLUMN = "category"
```

```python
# Make sure column names are clean and lowercase
df.columns = (
    df.columns
    .astype(str)
    .str.strip()
    .str.lower()
    .str.replace(" ", "_")
)
TEXT_COLUMN = "user_message"
LABEL_COLUMN = "intent"
RESPONSE_COLUMN = "bot_response"

# Display actual columns
print("Available columns:")
print(df.columns.tolist())

required_columns = [
    TEXT_COLUMN,
    LABEL_COLUMN,
    RESPONSE_COLUMN
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing columns: {missing_columns}"
    )

print("\nText column:", TEXT_COLUMN)
print("Label column:", LABEL_COLUMN)
print("Response column:", RESPONSE_COLUMN)

print("\nDataset shape:", df.shape)

print("\nSample data:")

display(
    df[
        [
            TEXT_COLUMN,
            LABEL_COLUMN,
            RESPONSE_COLUMN
        ]
    ].head(10)
)
```

```python

import re
import string

def clean_text(text):
    text = str(text).lower()
    text = re.sub(r"http\S+|www\S+|https\S+", "", text)
    
    text = re.sub(r"\S+@\S+", "", text)
    text = re.sub(r"\d+", "", text)

    
    text = text.translate(
        str.maketrans("", "", string.punctuation)
    )

    
    text = re.sub(r"\s+", " ", text).strip()

    return text


df["clean_text"] = df[TEXT_COLUMN].apply(clean_text)

df = df[df["clean_text"].str.strip() != ""].copy()

df = df.dropna(
    subset=[
        "clean_text",
        LABEL_COLUMN,
        RESPONSE_COLUMN
    ]
).reset_index(drop=True)

print("Dataset after cleaning:", df.shape)

display(
    df[
        [
            TEXT_COLUMN,
            "clean_text",
            LABEL_COLUMN
        ]
    ].head(10)
)
```

```python
print("Total unique intents:",
      df[LABEL_COLUMN].nunique())

print("\nIntent distribution:")

intent_counts = df[LABEL_COLUMN].value_counts()

display(intent_counts.to_frame("Count"))
```

```python
plt.figure(figsize=(12, 6))

sns.barplot(
    x=intent_counts.index,
    y=intent_counts.values
)

plt.title("Intent Distribution")
plt.xlabel("Intent")
plt.ylabel("Number of Messages")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
from sklearn.preprocessing import LabelEncoder

label_encoder = LabelEncoder()

df["label"] = label_encoder.fit_transform(
    df[LABEL_COLUMN].astype(str)
)

print("Number of classes:",
      len(label_encoder.classes_))

print("\nIntent Mapping:")

for number, intent in enumerate(
    label_encoder.classes_
):
    print(number, "->", intent)
```

```python
X = df["clean_text"]
y = df["label"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

class_counts = y.value_counts()

print("Minimum samples in any class:",
      class_counts.min())

if class_counts.min() >= 2:

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )
    print("Stratified split used.")

else:

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42
    )

    print("Regular split used because some intents have only one sample.")


print("\nTraining samples:", len(X_train))
print("Testing samples:", len(X_test))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(
    max_features=5000,
    ngram_range=(1, 2),
    sublinear_tf=True,
    min_df=1
)

X_train_tfidf = vectorizer.fit_transform(
    X_train
)

X_test_tfidf = vectorizer.transform(
    X_test
)

print("Training TF-IDF shape:",
      X_train_tfidf.shape)

print("Testing TF-IDF shape:",
      X_test_tfidf.shape)

print("\nTF-IDF vectorization completed successfully.")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=2000,
        random_state=42
    ),

    "Naive Bayes": MultinomialNB(),

    "Linear SVM": LinearSVC(
        random_state=42
    )
}

results = {}
trained_models = {}

for name, model in models.items():

    print("\nTraining:", name)

    model.fit(
        X_train_tfidf,
        y_train
    )

    predictions = model.predict(
        X_test_tfidf
    )

    accuracy = accuracy_score(
        y_test,
        predictions
    )

    results[name] = accuracy
    trained_models[name] = model

    print(
        "Accuracy:",
        round(accuracy * 100, 2),
        "%"
    )

print("\nAll models trained successfully.")
```

```python
results_df = pd.DataFrame({
    "Model": list(results.keys()),
    "Accuracy": list(results.values())
})

results_df["Accuracy (%)"] = (
    results_df["Accuracy"] * 100
)

display(results_df)
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=results_df,
    x="Model",
    y="Accuracy (%)"
)

plt.title(
    "AI Chatbot Model Accuracy Comparison"
)

plt.xlabel("Machine Learning Model")
plt.ylabel("Accuracy (%)")

plt.ylim(0, 100)

plt.xticks(rotation=15)

plt.tight_layout()
plt.show()
```

```python
best_model_name = max(
    results,
    key=results.get
)

best_model = trained_models[
    best_model_name
]

best_accuracy = results[
    best_model_name
]

print("Selected Model:")
print(best_model_name)

print(
    "\nAccuracy:",
    round(best_accuracy * 100, 2),
    "%"
)
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix
)

y_pred = best_model.predict(
    X_test_tfidf
)

print("Classification Report\n")
print("=" * 70)

print(
    classification_report(
        y_test,
        y_pred,
        labels=list(range(len(label_encoder.classes_))),
        target_names=label_encoder.classes_,
        zero_division=0
    )
)
```

```python
cm = confusion_matrix(
    y_test,
    y_pred,
    labels=list(range(len(label_encoder.classes_)))
)

plt.figure(figsize=(14, 10))

sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=label_encoder.classes_,
    yticklabels=label_encoder.classes_
)

plt.title(
    "Confusion Matrix - AI Chatbot"
)

plt.xlabel("Predicted Intent")
plt.ylabel("Actual Intent")

plt.xticks(rotation=45)
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```

```python
response_mapping = {}

for intent in df[LABEL_COLUMN].unique():

    responses = (
        df.loc[
            df[LABEL_COLUMN] == intent,
            RESPONSE_COLUMN
        ]
        .dropna()
        .astype(str)
        .tolist()
    )

    if len(responses) > 0:
        response_mapping[intent] = responses


print(
    "Number of intent response groups:",
    len(response_mapping)
)

print("\nSample responses:")

for intent, responses in list(
    response_mapping.items()
)[:5]:

    print("\nIntent:", intent)
    print("Response:", responses[0])
    
```

```python
import random

def chatbot(user_message):

    if not isinstance(user_message, str):
        return (
            "unknown",
            "Please enter a valid message."
        )

    if not user_message.strip():
        return (
            "unknown",
            "Please enter a message."
        )

    cleaned_message = clean_text(
        user_message
    )

    message_vector = vectorizer.transform(
        [cleaned_message]
    )

    predicted_label = best_model.predict(
        message_vector
    )[0]

    predicted_intent = (
        label_encoder.inverse_transform(
            [predicted_label]
        )[0]
    )

    possible_responses = response_mapping.get(
        predicted_intent,
        []
    )

    if possible_responses:

        response = random.choice(
            possible_responses
        )

    else:

        response = (
            "Sorry, I don't have a response "
            "for that question yet."
        )

    return predicted_intent, response
```

```python
test_messages = [
    "How do I add a course to my schedule?",
    "When is the library open?",
    "I forgot my student portal password.",
    "What are the exam dates?",
    "How do I connect to WiFi?"
]

for message in test_messages:

    intent, response = chatbot(
        message
    )

    print("User:", message)
    print("Predicted Intent:", intent)
    print("Bot:", response)
    print("-" * 70)
    
```

```python
def saved_chatbot(user_message):

    cleaned = clean_text(user_message)

    vector = test_vectorizer.transform(
        [cleaned]
    )

    predicted_label = test_model.predict(
        vector
    )[0]

    predicted_intent = test_encoder.inverse_transform(
        [predicted_label]
    )[0]

    responses = test_responses.get(
        predicted_intent,
        []
    )

    if responses:
        response = responses[0]
    else:
        response = "Sorry, I don't have an answer for that."

    return predicted_intent, response
```

```python
streamlit_code = r'''
import streamlit as st
import pickle
import re
import string
import random


# ==========================================
# PAGE CONFIG
# ==========================================

st.set_page_config(
    page_title="AI-Powered Chatbot",
    page_icon="🤖",
    layout="centered"
)


# ==========================================
# LOAD MODEL FILES
# ==========================================

with open("chatbot_model.pkl", "rb") as file:
    model = pickle.load(file)

with open("tfidf_vectorizer.pkl", "rb") as file:
    vectorizer = pickle.load(file)

with open("label_encoder.pkl", "rb") as file:
    label_encoder = pickle.load(file)

with open("response_mapping.pkl", "rb") as file:
    response_mapping = pickle.load(file)


# ==========================================
# TEXT CLEANING
# ==========================================

def clean_text(text):

    text = str(text).lower()

    text = re.sub(
        r"http\S+|www\S+|https\S+",
        "",
        text
    )

    text = re.sub(
        r"\S+@\S+",
        "",
        text
    )

    text = re.sub(
        r"\d+",
        "",
        text
    )

    text = text.translate(
        str.maketrans(
            "",
            "",
            string.punctuation
        )
    )

    text = re.sub(
        r"\s+",
        " ",
        text
    ).strip()

    return text


# ==========================================
# CHATBOT
# ==========================================

def chatbot(user_message):

    cleaned = clean_text(user_message)

    vector = vectorizer.transform(
        [cleaned]
    )

    predicted_label = model.predict(
        vector
    )[0]

    predicted_intent = label_encoder.inverse_transform(
        [predicted_label]
    )[0]

    responses = response_mapping.get(
        predicted_intent,
        []
    )

    if responses:
        response = random.choice(responses)
    else:
        response = (
            "Sorry, I don't have an answer "
            "for that question."
        )

    return predicted_intent, response


# ==========================================
# CUSTOM CSS
# ==========================================

st.markdown("""
<style>

.title {
    text-align: center;
    font-size: 38px;
    font-weight: bold;
    margin-bottom: 5px;
}

.subtitle {
    text-align: center;
    color: gray;
    margin-bottom: 25px;
}

</style>
""", unsafe_allow_html=True)


# ==========================================
# HEADER
# ==========================================

st.markdown(
    '<div class="title">🤖 AI-Powered Chatbot</div>',
    unsafe_allow_html=True
)

st.markdown(
    '<div class="subtitle">College Support Assistant</div>',
    unsafe_allow_html=True
)


# ==========================================
# SESSION STATE
# ==========================================

if "messages" not in st.session_state:
    st.session_state.messages = []


# ==========================================
# CHAT HISTORY
# ==========================================

for message in st.session_state.messages:

    with st.chat_message(
        message["role"]
    ):

        st.write(
            message["content"]
        )


# ==========================================
# USER INPUT
# ==========================================

user_input = st.chat_input(
    "Ask your question..."
)


if user_input:

    # User message

    st.session_state.messages.append({
        "role": "user",
        "content": user_input
    })

    with st.chat_message("user"):
        st.write(user_input)


    # Bot response

    intent, response = chatbot(
        user_input
    )

    st.session_state.messages.append({
        "role": "assistant",
        "content": response
    })

    with st.chat_message("assistant"):
        st.write(response)


# ==========================================
# SIDEBAR
# ==========================================

with st.sidebar:

    st.header("🤖 Chatbot")

    st.write(
        "AI-powered college support assistant."
    )

    st.divider()

    st.subheader("Features")

    st.write("✓ NLP")
    st.write("✓ TF-IDF")
    st.write("✓ Intent Classification")
    st.write("✓ Machine Learning")
    st.write("✓ Automated Responses")

    st.divider()

    if st.button("Clear Chat"):

        st.session_state.messages = []

        st.rerun()
'''


with open(
    "app.py",
    "w",
    encoding="utf-8"
) as file:

    file.write(streamlit_code)


print("app.py created successfully!")
```
